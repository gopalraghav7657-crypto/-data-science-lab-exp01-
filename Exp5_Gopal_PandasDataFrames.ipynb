```python
#Q28
school_df = pd.DataFrame({
    'Student': ['Aman', 'Ravi', 'Gopal', 'Neha'],
    'Class': [10, 10, 12, 12],
    'Section': ['A', 'B', 'A', 'B']
})

new_df = pd.DataFrame(
    school_df,
    columns=['Student', 'Class', 'Section', 'House']
)

print(new_df)

new_df['House'] = 'Red House'

print(new_df)
```

```python
#Q29
inventory_df = pd.DataFrame({
    'Product': ['Laptop', 'Mobile', 'Tablet', 'Mouse'],
    'Price': [50000, 20000, 30000, 500]
})

stock = pd.Series(
    [10, 20],
    index=[0, 2]
)

inventory_df['Stock'] = stock

print(inventory_df)

inventory_df = inventory_df.drop('Stock', axis=1)

print(inventory_df)
```
