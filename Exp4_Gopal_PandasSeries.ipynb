```python
#Q29
import pandas as pd
sales = pd.Series(
    [800, 450, 650, 300],
    index=['Laptop', 'Mouse', 'Keyboard', 'Monitor']
)
print("Values:")
print(sales.values)

print("\nIndex:")
print(sales.index)

print("\nLaptop Sales:")
print(sales['Laptop'])

print("\nSales above 500:")
print(sales[sales > 500])

print("\nIs Mouse present?")
print('Mouse' in sales.index)

sales_dict = sales.to_dict()

print("\nDictionary:")
print(sales_dict)
```
