```python
#Q35
import numpy as np

np.random.seed(42)

marks = np.random.randint(20, 101, size=(12, 4))

print("Student Marks:")
print(marks)

avg_student = marks.mean(axis=1)
print("Student averages:", avg_student)

test_mean = marks.mean(axis=0)
print("Test means:", test_mean)

test_std = marks.std(axis=0)
print("Test standard deviation:", test_std)

pass_all = np.all(marks >= 40, axis=1)
print("Students passing all tests:", np.sum(pass_all))

high_avg = np.where(avg_student >= 75)[0] + 1
print("Students with average >= 75:", high_avg)

print("Test with greatest variation:", np.argmax(test_std) + 1)
```

```python
#Q36
a = np.arange(21, 46).reshape(5, 5)

print("Original matrix:")
print(a)

middle = a[1:4, 1:4]

print("Central matrix:")
print(middle)

transposed = middle.T

print("Transpose:")
print(transposed)

product = middle @ transposed

print("Product:")
print(product)

row_sums = product.sum(axis=1)

print("Row sums:", row_sums)

maximum = product.max()

print("Maximum:", maximum)

positions = np.argwhere(product == maximum)

print("Maximum position:", positions)
```

```python
#Q37
qty = np.array([
    [10, 20, 15, 5],
    [18, 12, 20, 10],
    [25, 15, 10, 12],
    [20, 18, 22, 15],
    [16, 25, 18, 20],
    [22, 20, 16, 25]
])

original_prices = np.array([150, 240, 90, 320])

daily_revenue = qty @ original_prices

print("Daily revenue:")
print(daily_revenue)

product_revenue = (qty * original_prices).sum(axis=0)

print("Product revenue:")
print(product_revenue)

print("Highest revenue day:", np.argmax(daily_revenue) + 1)

total_quantity = qty.sum(axis=0)

print("Total quantity:")
print(total_quantity)

print("Best-selling product:", np.argmax(total_quantity) + 1)

discounted_prices = np.where(
    total_quantity > 100,
    original_prices * 0.90,
    original_prices
)

print("Discounted prices:", discounted_prices)

new_revenue = (qty * discounted_prices).sum()

print("New total revenue:", new_revenue)

print("Original prices:", original_prices)
```

```python
#Q38
np.random.seed(42)

original = np.random.normal(50, 12, 100)

outside = np.sum((original < 30) | (original > 70))

print("Values outside range:", outside)

cleaned = np.where(
    original < 30,
    30,
    np.where(original > 70, 70, original)
)

print("Cleaned data:")
print(cleaned)

print("Original mean:", np.mean(original))
print("Original variance:", np.var(original))
print("Original standard deviation:", np.std(original))
print("Original minimum:", np.min(original))
print("Original maximum:", np.max(original))

print("Cleaned mean:", np.mean(cleaned))
print("Cleaned variance:", np.var(cleaned))
print("Cleaned standard deviation:", np.std(cleaned))
print("Cleaned minimum:", np.min(cleaned))
print("Cleaned maximum:", np.max(cleaned))
```

```python
#Q39
branch_a = pd.Series({
    "Q1": 120,
    "Q2": 145,
    "Q4": 180
})

branch_b = pd.Series({
    "Q2": 110,
    "Q3": 135,
    "Q4": 155
})

branch_c = pd.Series({
    "Q1": 95,
    "Q3": 125,
    "Q4": 140
})

result = branch_a + branch_b + branch_c

print("Ordinary addition:")
print(result)

common = branch_a.index.intersection(
    branch_b.index
).intersection(branch_c.index)

print("Common quarters:", list(common))

combined = (
    branch_a[common]
    + branch_b[common]
    + branch_c[common]
)

print("Combined common earnings:")
print(combined)

zero_result = (
    branch_a.add(branch_b, fill_value=0)
    .add(branch_c, fill_value=0)
)

print("Missing values treated as zero:")
print(zero_result)
```

```python
#Q40
scores = {
    "R101": 78,
    "R102": 65,
    "R104": 92,
    "R105": 48
}

required = [
    "R105",
    "R103",
    "R101",
    "R106",
    "R104",
    "R102"
]

s = pd.Series(scores, index=required)

print("Original Series:")
print(s)

print("Missing records:")
print(s[s.isnull()])

valid = s[s.notnull() & (s >= 70)]

print("Valid scores:")
print(valid)

data = valid.to_dict()

print("Dictionary:")
print(data)

new_series = pd.Series(data)

print("Reconstructed Series:")
print(new_series)

print("Values match:", valid.equals(new_series))
```
