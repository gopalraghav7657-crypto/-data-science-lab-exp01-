```python
#Q4.4
sentence = "  Learning Python is fun  "

# (a) Remove leading and trailing spaces
clean_sentence = sentence.strip()

# (b) Count how many times 'n' appears
count_n = clean_sentence.count('n')

# (c) Split the sentence into a list of words
words = clean_sentence.split()

print("Without spaces:", clean_sentence)
print("Count of 'n':", count_n)
print("Words list:", words)
```
