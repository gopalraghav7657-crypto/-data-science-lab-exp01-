```python
#Q26
df1 = pd.read_csv("section_CA.csv")
df2 = pd.read_csv("section_CB.csv")
marks = pd.read_csv("exam_marks.csv")

roster = pd.concat([df1, df2], ignore_index=True)

result = pd.merge(roster, marks, on="Student_ID", how="left")

print(result[["Student_ID", "Student_Name", "Section",
              "Python_Marks", "DBMS_Marks"]])

print("Missing Marks:")
print(result[result["Python_Marks"].isnull() |
              result["DBMS_Marks"].isnull()])
```

```python
#Q27
df1 = pd.read_csv("marks_original.csv", index_col="Student_ID")
df2 = pd.read_csv("marks_updated.csv", index_col="Student_ID")

marks = df1["Marks"].combine_first(df2["Marks"])

students = pd.read_csv("students_master.csv", index_col="Student_ID")

result = pd.merge(students, marks, on="Student_ID", how="left")
result
```

```python
#28
df1 = pd.read_csv("marks_original.csv", index_col="Student_ID")
df2 = pd.read_csv("marks_updated.csv", index_col="Student_ID")

a = df1["Marks"].combine_first(df2["Marks"])
b = df1["Marks"].combine(df2["Marks"], np.fmax)

result = pd.DataFrame({
    "Combine_First": a,
    "Fmax": b
})

print(result.sort_index())

print("Different Student IDs:")
print(result[result["Combine_First"] != result["Fmax"]])
```

```python
#Q29
df = pd.read_csv("repeat_tests.csv")

result = df.pivot_table(
    index="Student_ID",
    columns="Subject",
    values="Marks",
    aggfunc=["mean", "max"]
)

print(result)

# Stack and restore
stacked = result.stack()
restored = stacked.unstack()

print("Restored Table:")
print(restored)
```
